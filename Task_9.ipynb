Here’s a Python method that calculates the daily interest correctly, accounting for actual days including the leap day, and applies payments to interest first, then principal:

from datetime import datetime

def audit_loan(principal, annual_rate, start_date, payments):
    """
    principal: initial loan amount
    annual_rate: annual interest rate (decimal, e.g., 0.12 for 12%)
    start_date: 'YYYY-MM-DD' string
    payments: list of tuples [(payment_date_str, payment_amount), ...]
    """
    balance = principal
    last_date = datetime.strptime(start_date, "%Y-%m-%d")
    audit_log = []

    for pay_date_str, payment in payments:
        pay_date = datetime.strptime(pay_date_str, "%Y-%m-%d")
        days_elapsed = (pay_date - last_date).days
        interest = balance * (annual_rate / 365) * days_elapsed
        interest_paid = min(payment, interest)
        principal_paid = payment - interest_paid
        balance -= principal_paid

        audit_log.append({
            "Period Start": last_date.strftime("%Y-%m-%d"),
            "Period End": pay_date.strftime("%Y-%m-%d"),
            "Days": days_elapsed,
            "Start Balance": round(balance + principal_paid, 2),
            "Interest Accrued": round(interest, 2),
            "Payment": payment,
            "Interest Paid": round(interest_paid, 2),
            "Principal Paid": round(principal_paid, 2),
            "End Balance": round(balance, 2)
        })

        last_date = pay_date

    return audit_log, round(balance, 2)

# Example usage:
principal = 100000
annual_rate = 0.12
start_date = "2024-01-01"
payments = [("2024-01-31", 5000), ("2024-02-29", 5000), ("2024-03-31", 5000)]

audit_log, final_balance = audit_loan(principal, annual_rate, start_date, payments)

for entry in audit_log:
    print(entry)
print("Final Balance:", final_balance)
