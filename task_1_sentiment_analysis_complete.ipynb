## Download missing NLTK resource and re-preprocess text

Download the `punkt_tab` NLTK resource, which was missing and causing a `LookupError` during tokenization. Then, re-apply the `preprocess_text` function to the 'text' column to create the 'processed_text' column.


**Reasoning**:
The previous code failed because the 'punkt_tab' NLTK resource was missing. I will first ensure all necessary NLTK resources are downloaded, then redefine the `preprocess_text` function, and finally apply it to the 'text' column to create the 'processed_text' column.



````python
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

try:
    nltk.data.find('tokenizers/punkt') # For word_tokenize, which calls sent_tokenize implicitly
except LookupError:
    nltk.download('punkt')

try:
    nltk.data.find('tokenizers/punkt_tab') # Explicitly download punkt_tab
except LookupError:
    nltk.download('punkt_tab')


def preprocess_text(text):
    # Convert to lowercase
    text = text.lower()
    # Remove punctuation
    text = ''.join([char for char in text if char not in string.punctuation])
    # Tokenize text
    tokens = word_tokenize(text)
    # Remove stop words
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    # Join tokens back into a string
    return ' '.join(filtered_tokens)

# Apply the preprocessing function to the 'text' column
df['processed_text'] = df['text'].apply(preprocess_text)

print("NLTK resources (stopwords, punkt, punkt_tab) are ready and text preprocessing completed. New 'processed_text' column created.")
display(df[['text', 'processed_text']].head())
````

**Reasoning**:
Having successfully preprocessed the text data, the next logical step is to prepare this processed text for topic modeling by converting it into a bag-of-words representation, which is a common requirement for algorithms like LDA.



````python
from gensim import corpora

# Tokenize the processed text column into a list of lists of words
documents = [text.split() for text in df['processed_text']]

# Create a dictionary from the tokenized documents
dictionary = corpora.Dictionary(documents)

# Filter out words that appear in less than no_below documents or in more than no_above fraction of documents
dictionary.filter_extremes(no_below=5, no_above=0.5)

# Create a bag-of-words corpus
corpus = [dictionary.doc2bow(doc) for doc in documents]

print("Text data prepared for topic modeling: dictionary and corpus created.")
print(f"Number of unique tokens in dictionary: {len(dictionary)}")
print(f"Number of documents in corpus: {len(corpus)}")
````

## Apply Latent Dirichlet Allocation (LDA)

### Subtask:
Apply an LDA topic modeling algorithm to the vectorized text to discover the underlying topics.


**Reasoning**:
Now that the text data is preprocessed and vectorized into a corpus and dictionary, I will import `LdaModel` from `gensim.models`, initialize an LDA model with 5 topics, and train it on the `corpus` using the `dictionary` as the `id2word` mapping.



````python
from gensim.models import LdaModel

# Set parameters for LDA model
num_topics = 5

# Initialize and train the LDA model
lda_model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, passes=10, random_state=42)

print(f"LDA model trained with {num_topics} topics.")
````

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Get the dominant topic for each document
def format_topics_sentences(ldamodel=None, corpus=None, texts=None):
    # Init output list to store topic data for each document
    topic_info_rows = []

    # Get main topic in each document
    for i, row_list in enumerate(ldamodel[corpus]):
        row = row_list[0] if ldamodel.per_word_topics else row_list
        row = sorted(row, key=lambda x: (x[1]), reverse=True)
        # Get the Dominant topic, Perc Contribution and Keywords for each document
        for j, (topic_num, prop_topic) in enumerate(row):
            if j == 0:  # => dominant topic
                wp = ldamodel.show_topic(topic_num)
                topic_keywords = ", ".join([word for word, prop in wp])
                topic_info_rows.append([int(topic_num), round(prop_topic,4), topic_keywords])
            else:
                break
    # Create the DataFrame once from the collected rows
    sent_topics_df = pd.DataFrame(topic_info_rows, columns=['Dominant_Topic', 'Perc_Contribution', 'Topic_Keywords'])

    # Add original text to the end of the output
    contents = pd.Series(texts)
    sent_topics_df = pd.concat([sent_topics_df, contents], axis=1)
    return sent_topics_df

# Call the function with the re-trained LDA model, corpus, and documents
df_topic_sents_keywords = format_topics_sentences(ldamodel=lda_model, corpus=corpus, texts=df['text'])

# Create the df_dominant_topic DataFrame
df_dominant_topic = df_topic_sents_keywords.reset_index()
df_dominant_topic.columns = ['Document_No', 'Dominant_Topic', 'Topic_Perc_Contrib', 'Keywords', 'Text']

print("Dominant topics assigned to documents.")
display(df_dominant_topic.head())

# Count the number of documents for each dominant topic
topic_counts = df_dominant_topic['Dominant_Topic'].value_counts().sort_index()

print("Distribution of dominant topics across documents:")
display(topic_counts)

# Visualize the distribution of dominant topics
plt.figure(figsize=(10, 6))
sns.barplot(x=topic_counts.index, y=topic_counts.values, hue=topic_counts.index, palette='viridis', legend=False)
plt.title('Distribution of Dominant Topics Across Documents')
plt.xlabel('Topic Number')
plt.ylabel('Number of Documents')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.75)
plt.show()
````
